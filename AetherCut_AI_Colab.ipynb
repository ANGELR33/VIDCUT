# 🎬 AetherCut AI - Cloud Studio (Google Colab)
Edita y genera videos con Inteligencia Artificial **100% en la nube**, sin consumir la memoria RAM de tu computadora ni gastar batería.

> **Instrucciones rápidas:**
> 1. Haz clic en **Entorno de ejecución > Cambiar tipo de entorno de ejecución** y selecciona **GPU T4** (o CPU estándar gratis).
> 2. Ejecuta los pasos 1, 2, 3 y 4 abajo en orden.
> 3. **Paso 2 (Persistencia con Google Drive):** Al activarlo, todos tus proyectos, videos editados y miniaturas se guardarán para siempre en tu Google Drive y **NUNCA se borrarán al reiniciar Colab**.
> 4. Haz clic en el enlace público generado en el paso 4 para abrir tu estudio.

```python
# @title 🚀 1. Descargar proyecto e instalar dependencias (~1 min)
import os
import subprocess

# Clonar o actualizar el repositorio
if not os.path.exists('VIDCUT'):
    !git clone https://github.com/ANGELR33/VIDCUT.git
    %cd VIDCUT
else:
    %cd VIDCUT
    !git pull origin master

# Instalar dependencias requeridas y tunelizador Cloudflare
!pip install -q --no-warn-conflicts -r requirements.txt pycloudflared
print('✅ Dependencias instaladas y código listo.')
```

```python
# @title 💾 2. Activar Persistencia Permanente con Google Drive (Recomendado)
# @markdown Conecta tu Google Drive para que tus proyectos, videos y claves NUNCA se borren al reiniciar Colab:
CONECTAR_GOOGLE_DRIVE = True # @param {type:'boolean'}

if CONECTAR_GOOGLE_DRIVE:
    try:
        from google.colab import drive
        import os
        import shutil

        # 1. Montar Google Drive si no está montado
        if not os.path.exists('/content/drive/MyDrive'):
            print('Montando Google Drive...')
            drive.mount('/content/drive')

        # 2. Carpeta persistente dedicada en tu Google Drive
        drive_storage = '/content/drive/MyDrive/AetherCut_AI_Storage'
        os.makedirs(drive_storage, exist_ok=True)
        for sub in ['projects', 'outputs', 'inputs', 'temp', 'assets']:
            os.makedirs(os.path.join(drive_storage, sub), exist_ok=True)

        # 3. Enlazar la carpeta storage local hacia Google Drive
        local_repo = '/content/VIDCUT'
        local_storage = os.path.join(local_repo, 'storage')

        if os.path.exists(local_storage) and not os.path.islink(local_storage):
            # Migrar datos locales previos si existieran
            for item in os.listdir(local_storage):
                s = os.path.join(local_storage, item)
                d = os.path.join(drive_storage, item)
                if not os.path.exists(d):
                    if os.path.isdir(s):
                        shutil.copytree(s, d)
                    else:
                        shutil.copy2(s, d)
            shutil.rmtree(local_storage)

        # Crear symlink absoluto
        os.system(f'ln -sfn "{drive_storage}" "{local_storage}"')
        os.environ['AETHERCUT_STORAGE_DIR'] = drive_storage

        # 4. Si existe .env guardado en Google Drive, restaurarlo automáticamente
        drive_env = os.path.join(drive_storage, '.env')
        local_env = os.path.join(local_repo, '.env')
        if os.path.exists(drive_env) and not os.path.exists(local_env):
            shutil.copy2(drive_env, local_env)
            print('🔑 Claves API restauradas automáticamente desde tu Google Drive.')

        print('✅ ¡Persistencia Permanente activada con éxito!')
        print(f'📁 Carpeta permanente: Google Drive > Mi unidad > AetherCut_AI_Storage')
        print('🎉 ¡Tus proyectos, escenas y videos NUNCA se borrarán, incluso si reinicias o cierras Colab!')
    except Exception as e:
        print(f'⚠️ No se pudo vincular Google Drive ({e}). Se continuará con almacenamiento temporal de la sesión.')
else:
    print('ℹ️ Persistencia en Drive omitida. Los proyectos solo vivirán mientras dure esta sesión de Colab.')
```

```python
# @title 🔑 3. Configurar Claves de API
# @markdown Pega tus claves de IA (Gemini, NVIDIA, GLM y KIMI):
GEMINI_API_KEY = '' # @param {type:'string'}
NVIDIA_API_KEY = '' # @param {type:'string'}
GLM_API_KEY = '' # @param {type:'string'}
GLM_FLASH_API_KEY = '' # @param {type:'string'}
KIMI_API_KEY = '' # @param {type:'string'}

# Si dejaste vacío, intenta leer de Colab Secrets o de Google Drive restaurado
try:
    from google.colab import userdata
    if not GEMINI_API_KEY: GEMINI_API_KEY = userdata.get('GEMINI_API_KEY')
    if not NVIDIA_API_KEY: NVIDIA_API_KEY = userdata.get('NVIDIA_API_KEY')
    if not GLM_API_KEY: GLM_API_KEY = userdata.get('GLM_API_KEY')
    if not GLM_FLASH_API_KEY: GLM_FLASH_API_KEY = userdata.get('GLM_FLASH_API_KEY')
    if not KIMI_API_KEY: KIMI_API_KEY = userdata.get('KIMI_API_KEY')
except Exception:
    pass

# Si aún está vacío, leer del .env local restaurado de Google Drive
if not GEMINI_API_KEY and os.path.exists('.env'):
    with open('.env', 'r', encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if line.startswith('GEMINI_API_KEY=') and not GEMINI_API_KEY:
                GEMINI_API_KEY = line.split('=', 1)[1]
            elif line.startswith('NVIDIA_API_KEY=') and not NVIDIA_API_KEY:
                NVIDIA_API_KEY = line.split('=', 1)[1]
            elif line.startswith('GLM_API_KEY=') and not GLM_API_KEY:
                GLM_API_KEY = line.split('=', 1)[1]
            elif line.startswith('GLM_FLASH_API_KEY=') and not GLM_FLASH_API_KEY:
                GLM_FLASH_API_KEY = line.split('=', 1)[1]
            elif line.startswith('KIMI_API_KEY=') and not KIMI_API_KEY:
                KIMI_API_KEY = line.split('=', 1)[1]

if not GEMINI_API_KEY:
    print('⚠️ Recuerda ingresar al menos tu GEMINI_API_KEY en la casilla.')
else:
    env_lines = [
        f'GEMINI_API_KEY={GEMINI_API_KEY}',
        f'NVIDIA_API_KEY={NVIDIA_API_KEY or ""}',
        f'GLM_API_KEY={GLM_API_KEY or ""}',
        f'GLM_FLASH_API_KEY={GLM_FLASH_API_KEY or ""}',
        f'KIMI_API_KEY={KIMI_API_KEY or ""}',
        'GEMINI_MODEL=gemini-flash-lite-latest',
    ]
    with open('.env', 'w', encoding='utf-8') as f:
        f.write('\n'.join(env_lines) + '\n')
    # Respaldar en Google Drive si está activo
    drive_storage = '/content/drive/MyDrive/AetherCut_AI_Storage'
    if os.path.exists(drive_storage):
        import shutil
        shutil.copy2('.env', os.path.join(drive_storage, '.env'))
        print('🔒 Copia de seguridad de tus claves guardada en Google Drive.')
    print('✅ Archivo .env configurado con todas tus claves.')
```

```python
# @title 🌐 4. Iniciar AetherCut AI y Generar Enlace HTTPS
# @markdown Ejecuta esta celda para iniciar el servidor y obtener tus enlaces de acceso.
import subprocess
import time
import sys
import os
from pycloudflared import try_cloudflare

# Asegurar que estamos en la carpeta del repositorio
if os.path.exists('/content/VIDCUT'):
    os.chdir('/content/VIDCUT')

# Asegurar persistencia activa con Google Drive si existe
drive_storage = '/content/drive/MyDrive/AetherCut_AI_Storage'
if os.path.exists(drive_storage):
    os.environ['AETHERCUT_STORAGE_DIR'] = drive_storage
    if not os.path.islink('/content/VIDCUT/storage'):
        os.system(f'ln -sfn "{drive_storage}" /content/VIDCUT/storage')

# Limpiar cualquier proceso antiguo para evitar bloqueos de puertos
subprocess.run(['pkill', '-9', '-f', 'uvicorn'], capture_output=True)
subprocess.run(['pkill', '-9', '-f', 'cloudflared'], capture_output=True)
subprocess.run(['git', 'pull', 'origin', 'master'], capture_output=True)
time.sleep(1)

print('Iniciando servidor de AetherCut AI en segundo plano...')
server_process = subprocess.Popen(
    ['uvicorn', 'server:app', '--host', '0.0.0.0', '--port', '8000'],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True
)

# Esperar a que el servidor inicialice
time.sleep(4)

# 1. Obtener enlace oficial directo de Google Colab (sin caídas ni problemas de DNS)
colab_url = None
try:
    from google.colab.output import eval_js
    colab_url = eval_js('google.colab.kernel.proxyPort(8000)')
except Exception:
    pass

# 2. Obtener enlace público Cloudflare (para celular o compartir)
cf_url = None
try:
    tunnel = try_cloudflare(port=8000)
    cf_url = getattr(tunnel, 'tunnel', str(tunnel))
except Exception:
    pass

print('\n' + '='*65)
print('🎉 ¡AETHERCUT AI ESTÁ ACTIVO EN LA NUBE!')
if colab_url:
    print('⭐ ENLACE DIRECTO (Google Oficial - Máxima velocidad):')
    print(f'👉 {colab_url}\n')
if cf_url:
    print('📱 ENLACE PÚBLICO (Cloudflare - Para celular o compartir):')
    print(f'👉 {cf_url}')
print('='*65 + '\n')
print('Tu laptop local consume 0% de memoria RAM.')

# Mostrar logs del servidor
try:
    while True:
        line = server_process.stdout.readline()
        if line:
            print(line.strip())
        time.sleep(0.2)
except KeyboardInterrupt:
    server_process.terminate()
    print('Servidor detenido.')
```
